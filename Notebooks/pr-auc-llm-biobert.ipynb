```python
import pandas as pd
import re
import numpy as np
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, label_binarize
from sklearn.metrics import (accuracy_score, classification_report, confusion_matrix, 
                             roc_curve, auc, precision_recall_curve, average_precision_score)
import seaborn as sns
import matplotlib.pyplot as plt
from transformers import BertTokenizer, BertForSequenceClassification

from torch.optim import AdamW
from imblearn.over_sampling import SMOTE
from torch.utils.data import DataLoader, TensorDataset

import warnings
warnings.filterwarnings("ignore", category=FutureWarning)

# Load BioBERT tokenizer and model
tokenizer = BertTokenizer.from_pretrained('dmis-lab/biobert-v1.1')
model = BertForSequenceClassification.from_pretrained('dmis-lab/biobert-v1.1', num_labels=40)
model.gradient_checkpointing_enable()

# Preprocess function
def preprocess_text(text):
    text = str(text).lower()
    text = re.sub(r'\([^)]*\)', '', text)
    text = re.sub('"', '', text)
    text = re.sub(r"'s\b", "", text)
    text = re.sub("[^a-zA-Z]", " ", text)
    return text.strip()

# Load and preprocess data
data = pd.read_csv('/kaggle/input/medicaltranscriptions/mtsamples.csv')
data.drop('Unnamed: 0', axis=1, inplace=True)
data['cleaned_transcription'] = data['transcription'].apply(preprocess_text)

# Tokenize
def encode_texts(texts):
    inputs = tokenizer(texts.tolist(), return_tensors='pt', padding=True, truncation=True, max_length=64)
    return inputs['input_ids'], inputs['attention_mask']

input_ids, attention_masks = encode_texts(data.cleaned_transcription)

# Encode labels
le = LabelEncoder()
Y = le.fit_transform(data.medical_specialty.values)

# Apply SMOTE
sm = SMOTE()
X_res, Y_res = sm.fit_resample(input_ids.numpy(), Y)

# Split data
X_train, X_val, Y_train, Y_val = train_test_split(X_res, Y_res, test_size=0.2, random_state=42)

# Convert to tensors
X_train_tensor = torch.tensor(X_train, dtype=torch.long)
Y_train_tensor = torch.tensor(Y_train, dtype=torch.long)
X_val_tensor = torch.tensor(X_val, dtype=torch.long)
Y_val_tensor = torch.tensor(Y_val, dtype=torch.long)

train_dataset = TensorDataset(X_train_tensor, Y_train_tensor)
val_dataset = TensorDataset(X_val_tensor, Y_val_tensor)

train_dataloader = DataLoader(train_dataset, batch_size=16, shuffle=True)
val_dataloader = DataLoader(val_dataset, batch_size=16, shuffle=False)

# Train with Early Stopping
def train_model_with_early_stopping(model, train_dataloader, val_dataloader, accumulation_steps=2, num_epochs=20, patience=2):
    optimizer = AdamW(model.parameters(), lr=5e-5)
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model.to(device)

    best_val_loss = float('inf')
    epochs_without_improvement = 0

    for epoch in range(num_epochs):
        model.train()
        total_loss = 0

        for batch in train_dataloader:
            optimizer.zero_grad()
            inputs, labels = batch
            inputs = inputs.to(device)
            labels = labels.to(device)
            outputs = model(input_ids=inputs, labels=labels)
            loss = outputs.loss
            total_loss += loss.item()
            loss.backward()
            optimizer.step()

        avg_loss = total_loss / len(train_dataloader)
        print(f"Epoch {epoch + 1}, Loss: {avg_loss:.4f}")

        # Validation
        model.eval()
        val_loss = 0.0
        val_preds, val_probs, val_labels = [], [], []

        with torch.no_grad():
            for batch in val_dataloader:
                inputs, labels = batch
                inputs = inputs.to(device)
                labels = labels.to(device)
                outputs = model(input_ids=inputs, labels=labels)
                val_loss += outputs.loss.item()
                logits = outputs.logits
                probs = torch.softmax(logits, dim=1)

                val_preds.append(torch.argmax(probs, dim=1).cpu().numpy())
                val_probs.append(probs.cpu().numpy())
                val_labels.append(labels.cpu().numpy())

        avg_val_loss = val_loss / len(val_dataloader)
        print(f"Validation Loss: {avg_val_loss:.4f}")

        # Early stopping
        if avg_val_loss < best_val_loss:
            best_val_loss = avg_val_loss
            epochs_without_improvement = 0
            torch.save(model.state_dict(), 'best_model.pth')
            print("Validation loss improved. Model saved.")
        else:
            epochs_without_improvement += 1
            print(f"No improvement for {epochs_without_improvement} epoch(s).")

        if epochs_without_improvement >= patience:
            print("Early stopping triggered.")
            break

        # Concatenate results
        val_preds = np.concatenate(val_preds)
        val_probs = np.concatenate(val_probs)
        val_labels = np.concatenate(val_labels)

        # Classification report
        print("\nClassification Report:")
        print(classification_report(val_labels, val_preds, target_names=le.classes_, digits=4, zero_division=0))

        # Confusion Matrix
        plt.figure(figsize=(17, 17))
        cm = confusion_matrix(val_labels, val_preds)
        sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=le.classes_, yticklabels=le.classes_)
        plt.title("Confusion Matrix")
        plt.ylabel("True Label")
        plt.xlabel("Predicted Label")
        plt.xticks(rotation=90)
        plt.yticks(rotation=0)
        plt.tight_layout()
        plt.show()

        # ROC AUC
        val_labels_bin = label_binarize(val_labels, classes=np.arange(len(le.classes_)))
        fpr, tpr, _ = roc_curve(val_labels_bin.ravel(), val_probs.ravel())
        roc_auc = auc(fpr, tpr)

        # PR-AUC Metrics
        precision_micro, recall_micro, _ = precision_recall_curve(val_labels_bin.ravel(), val_probs.ravel())
        pr_auc_micro = average_precision_score(val_labels_bin, val_probs, average='micro')
        pr_auc_macro = average_precision_score(val_labels_bin, val_probs, average='macro')
        pr_auc_weighted = average_precision_score(val_labels_bin, val_probs, average='weighted')

        # Micro PR Curve
        plt.figure(figsize=(8, 6))
        plt.plot(recall_micro, precision_micro, color='red', label=f'Micro PR-AUC = {pr_auc_micro:.4f}')
        plt.xlabel('Recall')
        plt.ylabel('Precision')
        plt.title('Micro Average PR-AUC Curve (BIOBERT)')
        plt.legend()
        plt.grid()
        plt.show()

        # Macro PR (reuse micro curve for simplicity in visualization)
        plt.figure(figsize=(8, 6))
        plt.plot(recall_micro, precision_micro, color='green', label=f'Macro PR-AUC = {pr_auc_macro:.4f}')
        plt.xlabel('Recall')
        plt.ylabel('Precision')
        plt.title('Macro Average PR-AUC Curve (BIOBERT)')
        plt.legend()
        plt.grid()
        plt.show()

        # Weighted PR
        plt.figure(figsize=(8, 6))
        plt.plot(recall_micro, precision_micro, color='blue', label=f'Weighted PR-AUC = {pr_auc_weighted:.4f}')
        plt.xlabel('Recall')
        plt.ylabel('Precision')
        plt.title('Weighted Average PR-AUC Curve (BIOBERT)')
        plt.legend()
        plt.grid()
        plt.show()

    print("Training complete.")

# Run training
train_model_with_early_stopping(model, train_dataloader, val_dataloader, num_epochs=25, patience=2)
```
